```python
import pandas as pd
import numpy as np
```

```python
# Load dataset
medical_df = pd.read_csv('KaggleV2-May-2016.csv')

# Display first 5 rows
medical_df.head()
```

```python
medical_df.tail()
```

```python
print("Dataset Shape:")
print(medical_df.shape)
```

```python
# This helps us understand all available features in the dataset.
print("Column Names:")
print(medical_df.columns)
```

```python
# The dtypes function shows the datatype of each column.
print("Data Types:")
print(medical_df.dtypes)
```

```python
# Dataset Information
medical_df.info()
```

```python
# This step identifies missing values in each column.
print("Missing Values:")
print(medical_df.isnull().sum())
```

```python
# Handle Missing Values
# Fill missing values for numerical columns
numerical_columns = medical_df.select_dtypes(include=['int64', 'float64']).columns

for column in numerical_columns:
    medical_df[column].fillna(medical_df[column].mean(), inplace=True)

# Fill missing values for categorical columns
categorical_columns = medical_df.select_dtypes(include=['object']).columns

for column in categorical_columns:
    medical_df[column].fillna(medical_df[column].mode()[0], inplace=True)

# Verify missing values again
print(medical_df.isnull().sum())
```

```python
# Statistical Analysis of Numerical Features
print("Statistical Summary:")
medical_df.describe()
```

```python
# Duplicate records can affect analysis accuracy.
print("Duplicate Rows:")
print(medical_df.duplicated().sum())
```

```python
# Removing duplicate rows keeps the dataset clean.
medical_df.drop_duplicates(inplace=True)
```

```python
# Convert Date Columns
medical_df['ScheduledDay'] = pd.to_datetime(medical_df['ScheduledDay'])
medical_df['AppointmentDay'] = pd.to_datetime(medical_df['AppointmentDay'])

print(medical_df[['ScheduledDay', 'AppointmentDay']].head())
```

```python
# Machine learning models cannot understand text values directly. So categorical columns are converted into numerical form.
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()

for column in medical_df.select_dtypes(include=['object']).columns:
    medical_df[column] = label_encoder.fit_transform(medical_df[column])
```

```python
# Verify that the dataset is clean and analysis-ready.
print("Final Dataset Information:")
medical_df.info()
```

```python
# Display the first rows of the cleaned dataset.
medical_df.head()
```

```python
# Save the processed dataset as a new CSV file.
medical_df.to_csv('cleaned_medical_appointment_data.csv', index=False)

print("Cleaned dataset saved successfully.")
```
